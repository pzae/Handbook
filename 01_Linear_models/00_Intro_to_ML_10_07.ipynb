# Ведение в машинное обучение

В повседневной жизни мы сталкиваемся с необходимостью принимать решения на основе информации из окружающего мира. Например, врач определяет болен пациент или здоров; банк решает, выдавать ли кредит человеку; интернет-магазин рекомендует товары. Все эти задачи объединяет общий принцип: мы хотим сделать предсказание на основе некоторой информации. Разберём этот принцип.
* Под "информацией" обычно подразумеваются различные характеристики или параметры, которые описывают интересующую нас ситуацию или предмет. В машинном обучении такие характеристики принято называть **признаками** или **фичами**, от английского *feature* - особенность, характеристика. Например, для задачи выдачи кредита такими признаками могут быть доход заявителя, его возраст, наличие собственности или трудовой стаж; в задаче о рекомендательном сервисе — это ваши прошлые покупки, частота посещения сайта, любимые категории товаров. Полный набор признаков называется **объектом**.
* Что касается предсказаний, то их принято называть **целевой переменной** или **таргетом**, от английского *target* - цель , так как получение верных предсказаний есть одна из основных целей машинного обучения. От типа значений, которые принимает целевая переменная, зависит формулировка задачи машинного обучения. Если целевая переменная принимает непрерывные числовые значения, например, стоимость жилья, температура воздуха или уровень дохода, то мы решаем задачу **регрессии**. Если же целевая переменная принимает ограниченное количество категориальных (дискретных) значений, например, "болен/здоров", "одобрено/отказано", "кошки/собаки", — это задача **классификации**.
* Осталось "на основе". Эта часть предполагает, что между признаками и целевой переменной есть некторая **связь**, которая позволяет по значениям признаков определить значение целевой переменной. В общем случае эту связь можно представить как **функцию**:

$$
f(\text{признаки}) = \text{таргет}
$$

Зная эту функцию, мы можем получить точное предсказание для любого **нового** набора признаков. Получение или описание этой связи есть ещё одна основная задача машинного обучения. На ней остановимся подробнее.

Как найти эту функцию? Какие функции бывают? Для начала выделим два больших семейства - параметрические и не параметрические. Не усложняя для вводной части можно сказать, что функции из первого семейства характеризуются тем, что их форма фиксирована заранее и зависит от некоторых внутренних параметров. В отличие от параметрических, непараметрические методы не фиксируют форму зависимости заранее, вместо этого они стараются гибко подстраиваться под данные. Приведём краткое сравнение. 

|                    | Параметрические              | Непараметрические                  |
|--------------------|-----------------------------|------------------------------------|
| Число параметров   | Фиксированное               | Растёт с количеством данных        |
| Форма функции      | Задана заранее              | Не задана, выбирается по данным    |
| Гибкость           | Меньшая                     | Более высокая                      |
| Пример             | Линейная регрессия          | Решающее дерево, kNN               |
| Необходимость данных | Меньше                     | Больше                             |

И скажем, что в данном хэндбуке нас интересуют именно параметрические функции. Параметры принято обозначать $\theta$ или $w$ от английского *weights* — веса. Ну а раз уж мы решили работать с параметрическими функциями, теперь нам нужно выбрать форму функции. Однако у внимательного читателя могут возникнуть вопросы - почему мы выбираем вид функции? А если истинный вид функции отличается от того, что мы выберем? В этот момент нам приходится признать, что мы не претендуем на поиск истины, а лишь на её **приближение**. Дело в том, что действительно, истинная зависимость между признаками и таргетом нам недоступна. Ближайший аналог - законы физики. Они могут сколько угодно точно описывать поведение объектов во Вселенной, но всегда остаётся "нечто", что нужно уточнить, чем нужно дополнить теорию. Так и мы выбираем путь нахождения приближения истинной функции, что неизбежно повлечёт за собой **ошибку**. Тогда наша связь между признаком и таргетом преобретает следующий вид:

$$
\hat{f}_{w}(\text{признаки}) + \varepsilon = \text{таргет}
$$

А связь между нашим приближением и истинной функцией:

$$
f(\text{признаки}) = \hat{f}_{w}(\text{признаки}) + \varepsilon
$$

При этом стоит отметить, что в эту $\epsilon$ можно заложить не только расхождение между приближением и истинной функцией, но и **неустранимый шум**, обусловленный случайными факторами, которые невозможно учесть в модели: погрешности измерений, внутренние состояния системы, которые нам неизвестны и прочее. Помимо таких неустранимых ошибок, в нашей оценке практически всегда присутсвует компонента ошибки, которую можно устранить, такая ситуация складывается, когда, например, наши параметры не достаточно хорошо описывают истинную функцию, но их можно изменить так, чтобы ошибка уменьшилась. Процесс подбора параметров и уменьшения ошибки называется **обучением** модели. Для корректной оценки ошибки нашего приближения вводится **функция потерь** или просто **лосс** от английского *loss* - ошибка. Тогда задача обучения формулируется как оптимизационная проблема: найти параметры модели, которые минимизируют функцию потерь, где параметры выступают в роли оптимизируемых переменных:

$$
w^* = \arg\min_{w} \mathcal{L}(w)
$$

где:
- $\mathcal{L}(w)$ — функция потерь, измеряющая отклонение предсказаний от реальных значений;
- $w^*$ — оптимальные параметры модели.

Подытоживая, можно заключить, что *решение задачи в машинном обучении всегда сопровождается вопросом: с какой точностью нужно это сделать?* Теперь, когда мы познакомились с азами машинного обучения, пора выбрать вид функции $\hat{f}_{w}(\text{признаки})$. 

# дописать часть про шум

В этот момент нам приходится признать, что мы не претендуем на поиск абсолютной истины, а лишь стремимся к её приближению. Истинная зависимость между признаками и таргетом нам недоступна. Ближе всего в этом смысле аналогия с законами физики: они могут, казалось бы, сколь угодно точно описывать поведение объектов во Вселенной, но всегда остаётся “что-то”, что требует уточнения или дополнения теории. Аналогично и мы в машинном обучении выбираем путь построения приближения истинной функции, что неизбежно влечёт за собой ошибку.

*Формальная запись*

Пусть существуют:
- Истинная (но неизвестная) функция зависимости между признаками и таргетом: $f(\text{признаки})$,
- Модель, которая аппроксимирует эту функцию: $\hat{f}_w(\text{признаки})$,
- Шум, обусловленный случайными факторами и погрешностями измерения: $\varepsilon$.

*Тогда наблюдаемые данные записываются так:*
$$
\text{таргет} = f(\text{признаки}) + \varepsilon
$$

*А аппроксимация истинной функции выглядит так:*
$$
f(\text{признаки}) = \hat{f}_w(\text{признаки}) + \delta
$$
где $\delta = f(\text{признаки}) - \hat{f}_w(\text{признаки})$ — это аппроксимационная ошибка, зависящая от того, насколько хорошо модель смогла подобрать параметры и от класса моделей вообще.

*Объединяя оба выражения:*
$$
\text{таргет} = \hat{f}_w(\text{признаки}) + \delta + \varepsilon
$$

где
- $\delta$ — ошибка аппроксимации (систематическая, потенциально устранимая, зависит от модели и параметров),
- $\varepsilon$ — шум (неустранимые случайные отклонения, непредсказуемые и независимые от модели).

---
*Комментарий по смыслу*

Таким образом, шум $\varepsilon$ включает в себя погрешности измерений, влияние неизвестных факторов и прочие случайные и неустранимые влияния. Ошибка аппроксимации $\delta$ появляется из-за недостатков выбранной модели (например, из-за того, что мы выбрали слишком простую функцию или не оптимизировали параметры).

Соответственно, полная ошибка предсказания состоит из суммы двух компонентов: той части, которую можно уменьшить (улучшаем модель — уменьшаем $\delta$), и той части, которую уменьшить невозможно ($\varepsilon$).

---
*Итоговая формула*

$$
\boxed{
\text{таргет} = \hat{f}_w(\text{признаки}) + \underbrace{\delta}_{\text{аппроксимационная ошибка}} + \underbrace{\varepsilon}_{\text{шум}}
}
$$

Здесь $\delta$ в принципе можно сделать сколь угодно малой при выборе подходящей модели и качественном обучении, тогда как $\varepsilon$ останется всегда — это неустранимая часть ошибки.
